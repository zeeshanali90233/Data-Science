# Probability of a Person to Die with Dengue

```python
df["Case_Fatality_Rate"] = (
    df["Dengue_Deaths"] / df["Dengue_Cases"]
) * 100
```

```python
print(df["Case_Fatality_Rate"].describe())
```

```python
cfr_region = (
    df.groupby("Region")
    .apply(
        lambda x: x["Dengue_Deaths"].sum()
        / x["Dengue_Cases"].sum() * 100,
        include_groups=False
    )
    .sort_values(ascending=False)
)

print(cfr_region)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=cfr_region.values,
    y=cfr_region.index
)

plt.title("Case Fatality Rate by Region")
plt.xlabel("Case Fatality Rate (%)")
plt.ylabel("Region")

plt.show()
```

Has the least Case Fatality Rate (CFR) - Seems Many reasons, May be have good hospitals or health services

# Executive Summary — Dengue Outbreak Analysis (2016–2020)

## Overview

This analysis examines **420 dengue records** covering **7 regions of Pakistan** from **2016 to 2020**, with a total of **177,135 reported dengue cases** and **1,165 reported deaths**. The objective is to identify major **temporal, regional, and severity patterns** in dengue cases and fatalities.

## Key Findings

### 📈 1. 2019 Recorded the Highest Number of Dengue Cases

Among the five years analyzed, **2019 experienced the highest number of reported dengue cases**, indicating that it was the most significant outbreak year within the dataset. This makes 2019 a particularly important year for understanding dengue trends and potential contributing factors.

### 🗺️ 2. Sindh Recorded the Highest Overall Cases

**Sindh recorded the highest total number of dengue cases**, followed closely by **Punjab**. The relatively small difference between the two suggests that both regions experienced substantial dengue activity during the study period.

However, an important regional pattern appears in **2019**: despite Sindh having the highest overall cases across the complete dataset, **Punjab recorded the highest number of cases in 2019**. This indicates that regional rankings can change considerably depending on the year being analyzed.

### ⚰️ 3. Sindh Had the Highest Number of Reported Deaths

Sindh also recorded the **highest total number of dengue deaths**, with Punjab again relatively close behind. This suggests that the regions with the greatest dengue burden also experienced a substantial share of dengue-related fatalities.

### 🦟 4. August Was the Peak Month

**August recorded the highest number of dengue cases** among the months analyzed. This indicates a potentially important **seasonal pattern**, with dengue activity increasing during this period.

The finding may be useful for planning **early surveillance, awareness campaigns, vector-control activities, and healthcare preparedness** before the expected peak.

### 📊 5. Dengue Deaths Are Strongly Associated With Dengue Cases

The analysis shows a **strong positive relationship between dengue cases and dengue deaths**. In general, periods or regions with more reported dengue cases also tend to report more deaths.

This suggests that controlling the overall number of infections may be an important factor in reducing the absolute number of dengue-related deaths.

### 🏥 6. Balochistan Had the Highest Case Fatality Rate

While **Sindh recorded the highest total number of dengue deaths**, **Balochistan had the highest Case Fatality Rate (CFR)** among all regions. This means that, relative to its total number of reported dengue cases, Balochistan recorded a **higher proportion of deaths** than the other regions.

This is an important distinction: **the region with the most deaths is not necessarily the region with the highest fatality rate**. Balochistan's higher CFR may indicate differences in disease severity, healthcare access, timely diagnosis and treatment, reporting practices, or other regional factors. However, the available dataset does not contain enough information to determine the exact cause.

Therefore, Balochistan's high CFR warrants **further investigation into healthcare accessibility, treatment availability, early detection, and other factors that may influence dengue outcomes**.


## Executive Conclusion

Overall, the analysis highlights **2019 as the most severe year**, **Sindh and Punjab as the two regions with the greatest dengue burden**, and **August as the peak month for reported cases**. The strong relationship between cases and deaths further emphasizes the importance of reducing dengue transmission and preparing healthcare systems ahead of peak periods.

The regional differences in **Case Fatality Rate** also demonstrate that the number of cases alone does not fully describe the impact of an outbreak. Further analysis incorporating **healthcare capacity, population size, climate/weather conditions, prevention measures, and treatment access** would be required to explain why fatality rates differ between regions.
