# Urbanization Rate

```python
urbanization = (
    rural_urban["URBAN_POPULATION"]
    / (
        rural_urban["RURAL_POPULATION"]
        + rural_urban["URBAN_POPULATION"]
    )
) * 100

urbanization = urbanization.sort_values(ascending=False)

print(urbanization)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=urbanization.values,
    y=urbanization.index
)

plt.title("Urbanization Rate by Province")
plt.xlabel("Urban Population (%)")
plt.ylabel("Province")

plt.show()
```

## Population Density

```python
df["POPULATION_DENSITY"] = (
    df["TOTAL_POPULATION"]
    / df["AREA (SQ.KM)"]
)
```

```python
print(
    df[
        [
            "PROVINCE",
            "DISTRICT",
            "SUB DIVISION",
            "AREA (SQ.KM)",
            "TOTAL_POPULATION",
            "POPULATION_DENSITY"
        ]
    ]
    .sort_values(
        "POPULATION_DENSITY",
        ascending=False
    )
    .head(10)
)
```

```python
province_density = (
    df.groupby("PROVINCE")
    .agg({
        "TOTAL_POPULATION": "sum",
        "AREA (SQ.KM)": "sum"
    })
)

province_density["POPULATION_DENSITY"] = (
    province_density["TOTAL_POPULATION"]
    / province_density["AREA (SQ.KM)"]
)

province_density = province_density.sort_values(
    "POPULATION_DENSITY",
    ascending=False
)

print(province_density)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=province_density["POPULATION_DENSITY"],
    y=province_density.index
)

plt.title("Population Density by Province")
plt.xlabel("People per sq. km")
plt.ylabel("Province")

plt.show()
```

```python
df["POPULATION_1998_TOTAL"] = (
    df["POPULATION 1998 (RURAL)"] +
    df["POPULATION 1998 (URBAN)"]
)
```

```python
df["TOTAL_POPULATION"] = (
    df["ALL SEXES (RURAL)"] +
    df["ALL SEXES (URBAN)"]
)
```

```python
df["POPULATION_INCREASE"] = (
    df["TOTAL_POPULATION"]
    - df["POPULATION_1998_TOTAL"]
)
```

```python
df["POPULATION_GROWTH_PERCENT"] = (
    (df["TOTAL_POPULATION"] - df["POPULATION_1998_TOTAL"])
    / df["POPULATION_1998_TOTAL"]
) * 100
```

```python
top_growth = (
    df[
        [
            "PROVINCE",
            "DISTRICT",
            "SUB DIVISION",
            "POPULATION_1998_TOTAL",
            "TOTAL_POPULATION",
            "POPULATION_GROWTH_PERCENT"
        ]
    ]
    .sort_values(
        "POPULATION_GROWTH_PERCENT",
        ascending=False
    )
    .head(10)
)

print(top_growth)
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_growth,
    x="POPULATION_GROWTH_PERCENT",
    y="SUB DIVISION"
)

plt.title("Top 10 Fastest-Growing Sub-Divisions (1998–2017)")
plt.xlabel("Population Growth (%)")
plt.ylabel("Sub-Division")

plt.show()
```

```python
lowest_growth = (
    df[
        [
            "PROVINCE",
            "DISTRICT",
            "SUB DIVISION",
            "POPULATION_1998_TOTAL",
            "TOTAL_POPULATION",
            "POPULATION_GROWTH_PERCENT"
        ]
    ]
    .sort_values(
        "POPULATION_GROWTH_PERCENT"
    )
    .head(10)
)

print(lowest_growth)
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=lowest_growth,
    x="POPULATION_GROWTH_PERCENT",
    y="SUB DIVISION"
)

plt.title("Top 10 Slowest-Growing Sub-Divisions (1998–2017)")
plt.xlabel("Population Growth (%)")
plt.ylabel("Sub-Division")

plt.show()
```

```python
province_growth = (
    df.groupby("PROVINCE")
    .agg(
        Population_1998=(
            "POPULATION_1998_TOTAL",
            "sum"
        ),
        Population_Current=(
            "TOTAL_POPULATION",
            "sum"
        )
    )
)

province_growth["Growth_Percent"] = (
    (
        province_growth["Population_Current"]
        - province_growth["Population_1998"]
    )
    / province_growth["Population_1998"]
) * 100

province_growth = province_growth.sort_values(
    "Growth_Percent",
    ascending=False
)

print(province_growth)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=province_growth["Growth_Percent"],
    y=province_growth.index
)

plt.title("Population Growth by Province (1998–2017)")
plt.xlabel("Population Growth (%)")
plt.ylabel("Province")

plt.show()
```

```python
population_comparison = province_growth[
    [
        "Population_1998",
        "Population_Current"
    ]
]
```

```python
population_comparison.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Population Comparison: 1998 vs Current Census")
plt.xlabel("Province")
plt.ylabel("Population")

plt.xticks(rotation=45)

plt.show()
```

```python
df["RURAL_GROWTH_PERCENT"] = (
    (
        df["ALL SEXES (RURAL)"]
        - df["POPULATION 1998 (RURAL)"]
    )
    / df["POPULATION 1998 (RURAL)"]
) * 100
```

```python
rural_growth = df.groupby("PROVINCE").agg(
    Rural_1998=("POPULATION 1998 (RURAL)", "sum"),
    Rural_Current=("ALL SEXES (RURAL)", "sum")
)

rural_growth["Growth_Percent"] = (
    (
        rural_growth["Rural_Current"]
        - rural_growth["Rural_1998"]
    )
    / rural_growth["Rural_1998"]
) * 100

print(rural_growth)
```

```python
df["URBAN_GROWTH_PERCENT"] = (
    (
        df["ALL SEXES (URBAN)"]
        - df["POPULATION 1998 (URBAN)"]
    )
    / df["POPULATION 1998 (URBAN)"]
) * 100
```

```python
urban_growth = df.groupby("PROVINCE").agg(
    Urban_1998=("POPULATION 1998 (URBAN)", "sum"),
    Urban_Current=("ALL SEXES (URBAN)", "sum")
)

urban_growth["Growth_Percent"] = (
    (
        urban_growth["Urban_Current"]
        - urban_growth["Urban_1998"]
    )
    / urban_growth["Urban_1998"]
) * 100

print(urban_growth)
```

```python
growth_comparison = pd.DataFrame({
    "Rural Growth (%)": rural_growth["Growth_Percent"],
    "Urban Growth (%)": urban_growth["Growth_Percent"]
})

print(growth_comparison)
```

```python
growth_comparison.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Rural vs Urban Population Growth (1998–2017)")
plt.xlabel("Province")
plt.ylabel("Growth (%)")

plt.xticks(rotation=45)

plt.show()
```

```python
df["URBANIZATION_1998"] = (
    df["POPULATION 1998 (URBAN)"]
    / df["POPULATION_1998_TOTAL"]
) * 100
```

```python
df["URBANIZATION_CURRENT"] = (
    df["ALL SEXES (URBAN)"]
    / df["TOTAL_POPULATION"]
) * 100
```

```python
df["URBANIZATION_CHANGE"] = (
    df["URBANIZATION_CURRENT"]
    - df["URBANIZATION_1998"]
)
```

```python
urbanization_province = df.groupby("PROVINCE").agg(
    Population_1998=(
        "POPULATION_1998_TOTAL",
        "sum"
    ),
    Urban_1998=(
        "POPULATION 1998 (URBAN)",
        "sum"
    ),
    Population_Current=(
        "TOTAL_POPULATION",
        "sum"
    ),
    Urban_Current=(
        "ALL SEXES (URBAN)",
        "sum"
    )
)

urbanization_province["Urbanization_1998"] = (
    urbanization_province["Urban_1998"]
    / urbanization_province["Population_1998"]
) * 100

urbanization_province["Urbanization_Current"] = (
    urbanization_province["Urban_Current"]
    / urbanization_province["Population_Current"]
) * 100

urbanization_province["Urbanization_Change"] = (
    urbanization_province["Urbanization_Current"]
    - urbanization_province["Urbanization_1998"]
)

print(urbanization_province)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=urbanization_province.reset_index(),
    x="PROVINCE",
    y="Urbanization_Change"
)

plt.title("Change in Urbanization Rate (1998–2017)")
plt.xlabel("Province")
plt.ylabel("Change in Urbanization (%)")

plt.xticks(rotation=45)

plt.show()
```
